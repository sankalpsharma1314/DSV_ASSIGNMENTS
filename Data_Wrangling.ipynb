```python
#QU26
import pandas as pd

ca = pd.read_csv('section_CA.csv')
cb = pd.read_csv('section_CB.csv')
marks = pd.read_csv('exam_marks.csv')

roster = pd.concat([ca, cb], axis=0, ignore_index=True)

df = pd.merge(roster, marks, on='Student_ID', how='left')

res = df[['Student_ID', 'Student_Name', 'Section', 'Python_Marks', 'DBMS_Marks']]

missing = res[res[['Python_Marks', 'DBMS_Marks']].isna().any(axis=1)]

print(res)
print("\nStudents with missing exam marks:")
print(missing)
```

```python
#QU27
import pandas as pd

orig = pd.read_csv('marks_original.csv').set_index('Student_ID')
upd = pd.read_csv('marks_updated.csv').set_index('Student_ID')
master = pd.read_csv('students_master.csv')

comb = orig.combine_first(upd).reset_index()

res = pd.merge(master, comb[['Student_ID', 'Marks']], on='Student_ID', how='left')

print(res)
```

```python
#QU28
import pandas as pd
import numpy as np

orig = pd.read_csv('marks_original.csv').set_index('Student_ID')
upd = pd.read_csv('marks_updated.csv').set_index('Student_ID')

res_a = orig.combine_first(upd)
res_b = orig.combine(upd, np.fmax)

comp = pd.DataFrame({
    'combine_first': res_a['Marks'],
    'combine_fmax': res_b['Marks']
})

diff_ids = comp[comp['combine_first'] != comp['combine_fmax']].dropna(how='all').index.tolist()

print(comp)
print("\nStudent_IDs where results differ:", diff_ids)
```

```python
#QU29
import pandas as pd

rt = pd.read_csv('repeat_tests.csv')

pt = pd.pivot_table(
    rt,
    index='Student_ID',
    columns='Subject',
    values='Marks',
    aggfunc=['mean', 'max']
)

st = pt.stack(future_stack=True)
restored = st.unstack()

print("Original Pivot Table:\n", pt)
print("\nStacked (Hierarchical Index):\n", st)
print("\nRestored Pivot Table:\n", restored)
```
