### 29. Write a short Pandas program that creates a labelled Series for four products and sales, displays its values and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a dictionary.

```python
import pandas as pd

# 1. Create Series
sales = pd.Series([450, 620, 300, 850], index=['Laptop', 'Phone', 'Tablet', 'Monitor'])

# 2. Values and Index
print('Values:', sales.values)
print('Index:', sales.index)

# 3. Select one label
print('Phone sales:', sales['Phone'])

# 4. Filter > 500
print('Sales > 500:\n', sales[sales > 500])

# 5. Check if label exists
print('Is Tablet present?', 'Tablet' in sales)

# 6. Convert to dict
print('As dict:', sales.to_dict())
```
